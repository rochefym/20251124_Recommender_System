### Vector Embedding

```python
from langchain_ollama import OllamaEmbeddings

import faiss
from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore
```

```python
embeddings = OllamaEmbeddings(model='nomic-embed-text:v1.5', base_url="http://localhost:11434")
```

```python
#Create a dummy vector for initial store in the database
single_vector = embeddings.embed_query("this is some text data")
single_vector
```

```python
index = faiss.IndexFlatL2(len(single_vector))
index.ntotal, index.d
```

```python
#Create vectore database & save locally
vector_store = FAISS(
    embedding_function=embeddings,
    index=index,
    docstore=InMemoryDocstore(),
    index_to_docstore_id={}
)
```

```python
ids = vector_store.add_documents(documents=chunks)
```

```python
vector_store.index_to_docstore_id
len(ids)
```

```python
# save vector database locally
db_name = "dietary_reference_intakes"
vector_store.save_local(db_name)
```

```python
# load vector database
db_name = "dietary_reference_intakes"

new_vector_store = FAISS.load_local(db_name, embeddings=embeddings, allow_dangerous_deserialization=True)
len(new_vector_store.index_to_docstore_id)
```

<br>

### Retrieval

```python
# (1) SIMILARITY SEARCH
question = "what is the maximum amount of carbohydrates a 40 year old adult can take daily?"
docs = vector_store.search(query=question, search_type='similarity')

for doc in docs:
    print(doc.page_content)
    print("\n\n")
```

```python
# (2) MMR = Maximal Marginal Relevance

retriever = vector_store.as_retriever(search_type="mmr", search_kwargs = {'k': 16, 
                                                                          'fetch_k': 100,
                                                                          'lambda_mult': 1})

question = "what is the maximum amount of carbohydrates a 40 year old adult can take daily?"
docs = retriever.invoke(question)

for doc in docs:
    print(doc.page_content)
    print("\n\n")
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are a professional clinical nutritionist specializing in elderly care in Taiwan.

Use the retrieved context below to support your reasoning.
- If context provides relevant data, use it directly.
- If it lacks exact values, give only safe, general suggestions.
- Do NOT make up numbers or conversions.
- Be concise, clear, and compassionate.

Question:
{question}

Context:
{context}

Please provide:
1. **Analysis** — summarize the nutritional content and adequacy.
2. **Suggestions** — what can be improved or balanced.
3. **Recommendations** — practical next steps for elderly dietary care.
""")

retriever = new_vector_store.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 8,          # fewer but higher-quality chunks
        "fetch_k": 50,   # how many to initially fetch before filtering
        "lambda_mult": 0.7  # balance between diversity and relevance
    }
)

def format_docs(docs):
    return "\n\n".join([d.page_content for d in docs])

from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | model
    | StrOutputParser()
)

question = (
    "A patient named Ronny Chen is a 76-year-old man. "
    "He is having 320 grams of stir-fried seasonal vegetables and "
    "220 grams of cabbage soup for his meal. "
    "Can you provide (1) an analysis, (2) suggestions, and (3) recommendations?"
)

output = rag_chain.invoke(question)
print(output)
```
